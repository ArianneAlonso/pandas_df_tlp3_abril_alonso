```python
import pandas as pd
import sqlite3

#Aqui agrega la conexión a la base de datos 
conn =sqlite3.connect('Salaries.sqlite')

#Crea el dataframe con los datos del datasets
df = pd.read_sql("SELECT * FROM salaries", conn)

#cerrar la conexión a la base de datos
conn.close()

print(df.head())
```

1. **Top 10 empleados con mayor salario total (incluyendo beneficios)**
Extrae e imprime un top 10 de expleados con mayor salario incluyendo los beneficios 


```python
#todo
diez_mejores_empleados = df.sort_values(by='TotalPayBenefits', ascending=False).head(10)

#sort_values creo qe es para ordenar, en ese caso es de forma descendente

print(diez_mejores_empleados[['EmployeeName', 'TotalPayBenefits']])
```

2. **Filtrar empleados con más de 50,000 en horas extra, la columna a tratar es: (OvertimePay)**

```python
#todo

df['OvertimePay'] = pd.to_numeric(df['OvertimePay'], errors='coerce')

empleados_horas_extra = df[df['OvertimePay'] > 50000]

print(empleados_horas_extra[['EmployeeName', 'JobTitle', 'OvertimePay']])
```

3. **Contar cuántos empleados únicos hay por año**

```python
#todo

empleados_unicos = df.groupby('Year')['EmployeeName'].nunique()
#groupby agrupa datos, en este caso años
#.nunique() cuenta cuantos empleados sin repetir hay en cada año, en este caso con ayuda de la columna de nombres

print(empleados_unicos)
```

4. **Ver cuántos cargos únicos (JobTitle) existen y los 5 más comunes**

```python
#todo

jobtitle = df['JobTitle'].nunique()
print("cargos unicos: ", jobtitle)

jobtitle_comunes = df['JobTitle'].value_counts().head(5)
print("cargo mas comunes: ", jobtitle_comunes)
```

5. **Mostrar el salario total (TotalPay) promedio por año**

```python
#todo
df['TotalPay'] = pd.to_numeric(df['TotalPay'], errors='coerce')

salario_promedio = df.groupby('Year')['TotalPay'].mean()

print(salario_promedio)
```
